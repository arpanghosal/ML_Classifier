```python
import numpy as np
import uproot
import pandas as pd
import matplotlib.pyplot as plt
import mplhep

import seaborn as sns
```

```python
data = np.load('roc_data_70.npz')
fpr_70 = data['fpr']
tpr_70 = data['tpr']
thresholds_70 = data['thresholds']
AUC_70 = data['auc']
print(f"Loaded data: AUC = {AUC_70}")


data = np.load('roc_data_85.npz')
fpr_85 = data['fpr']
tpr_85 = data['tpr']
thresholds_85 = data['thresholds']
AUC_85 = data['auc']
print(f"Loaded data: AUC = {AUC_85}")




data = np.load('roc_data_85_reweighted.npz')
fpr_85_reweighted = data['fpr']
tpr_85_reweighted = data['tpr']
thresholds_85_reweighted = data['thresholds']
AUC_85_reweighted = data['auc']
print(f"Loaded data: AUC = {AUC_85_reweighted}")




data = np.load('roc_data_85_reweighted_perprocess.npz')
fpr_85_reweighted_perprocess = data['fpr']
tpr_85_reweighted_perprocess = data['tpr']
thresholds_85_reweighted_perprocess = data['thresholds']
AUC_85_reweighted_perprocess = data['auc']
print(f"Loaded data: AUC = {AUC_85_reweighted_perprocess}")


data = np.load('roc_data_85_reweighted_perprocess_3class.npz')
fpr_85_reweighted_perprocess_3class = data['fpr']
tpr_85_reweighted_perprocess_3class = data['tpr']
thresholds_85_reweighted_perprocess_3class = data['thresholds']
AUC_85_reweighted_perprocess_3class = data['auc']
print(f"Loaded data: AUC = {AUC_85_reweighted_perprocess_3class}")

```

```python
mplhep.style.use('ATLAS')    
```

```python
# AUC's comparison:


#plt.plot(fpr_85,tpr_85,label=F'AUC (training with WP85) = {AUC_85:.6f}', color='blue')
#plt.plot(fpr_85_reweighted,tpr_85_reweighted,label=F'AUC (training with WP85 reweighted to WP70) = {AUC_85_reweighted:.6f}', color='red')
plt.plot(fpr_85_reweighted_perprocess,tpr_85_reweighted_perprocess,label=F'AUC (trained with WP85, reweighted to WP70 per process (2-class, Model A)) = {AUC_85_reweighted_perprocess:.6f}', color='cyan')
plt.plot(fpr_85_reweighted_perprocess_3class,tpr_85_reweighted_perprocess_3class,label=F'AUC (trained with WP85, reweighted to WP70 per process (3-class, Model B)) = {AUC_85_reweighted_perprocess_3class:.6f}', color='magenta')
plt.plot(fpr_70,tpr_70,label=F'AUC (trained with WP70 (baseline model))  = {AUC_70:.6f}', color='orange')


plt.legend(loc = 'lower right', fontsize=8)
plt.xlim(0,1)
plt.ylim(0,1.15)
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
mplhep.atlas.label('Internal', data=False)
```
